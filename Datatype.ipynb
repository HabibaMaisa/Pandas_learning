## Numeric Data Type in Pandas

```python
import pandas as pd
```

```python
#creating a Dataframe with integer data
data={
    'Age':[25,30,26,28,20],
    'salary':[2334,1434,13434,3434,2343]}
df=pd.DataFrame(data)
df
```

```python
df.info()
```

```python
df.dtypes
```

```python
df['Age']=df['Age'].astype('int32')
df['salary']=df['salary'].astype('int32')
df.dtypes
```

## Object datatype

```python
# Create a dictionary with mixed data types
data={'Name':['Rr','Mm','Aa','Ss'],
      'Age':[25,23,24,26],
      'Occupation' :['Data Sci','Software','Teacher','engineer'],
      'Salary':[2344,34653,23453,56234]}
data
```

```python
df=pd.DataFrame(data)
print(df.dtypes)
```

```python
# Create a dictionary with mixed data types
data={'Name':['Rr','Mm','Aa','Ss'],
      'Age':[25,23,24,26],
      'Occupation' :['Data Sci','Software','Teacher','engineer'],
      'Salary':['BDT 2344','BDT 34653','BDT 23453','BDT 56234']}
data
```

## Datetime

```python
#create a list of dates 
date=['2026-04-16','2026-04-17','2026-04-18']
df1=pd.DataFrame({'Date':date})
df1['Date'].dtype
```

```python
timestamps=pd.to_datetime(date)
df2=pd.DataFrame({'Date2':timestamps})
df2['Date2'].dtype
```
