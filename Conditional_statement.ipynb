## Conditional Satements

```python
import pandas as pd
```

```python
df=pd.read_csv('mental_health.csv')
df.head()
```

```python
df['Daily_Screen_Time'].max()
```

```python
df['Daily_Screen_Time']>8
```

```python
df[df['Daily_Screen_Time']>8]
```

```python
#subseting rows
df[df['Occupation']=='Student']
```

```python
df[(df['Occupation']=='Student') | (df['Occupation']=='Unemployed')]
```

```python
#subsetting using isin()
df['Occupation'].isin(['Student','Unemployed'])
```

## Common Q&A

```python
# show occupation= student and daily sccren time 2
df[(df['Occupation']=='Student')& (df['Daily_Screen_Time']==2)]
```

```python
#show occupation where daily screen time is >4

df['Occupation'][df['Daily_Screen_Time']>4]
```

```python
# show the raws with minimum Daily_Screen_Time
df[df['Daily_Screen_Time']==df['Daily_Screen_Time'].min()]
```

```python
#show age and occupation where dailly screen time is greater than 4
df[['Age','Occupation']][df['Daily_Screen_Time']>4]
```
